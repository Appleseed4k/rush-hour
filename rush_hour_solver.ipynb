# Batch AND-OR solving

Reads every puzzle out of a `rush_nw.txt`-format file into a distance -> states dict via `rush_hour_lib.read_puzzles`, solves each with `rush_hour_and_or.solve`, and saves a `rush_hour_lib.visualize` trace (one PNG per move plus a distance-to-goal plot) into its own subfolder under `solve_attempts/`.

```python
import os
import shutil

import rush_hour_and_or as ao
import rush_hour_lib as lib
```

```python
INPUT_PATH = "data/rush_nw_training.txt"
OUTPUT_DIR = "solve_attempts"
MAX_STEPS = 200
```

```python
puzzles = lib.read_puzzles(INPUT_PATH)
sum(len(states) for states in puzzles.values())
```

```python
def solve_puzzle(state, max_steps=MAX_STEPS):
    """Repeatedly calls ao.solve until it returns a full move list (solved) or
    gives up (a dead end, or max_steps runs out). Returns the move list, or
    None if unsolved."""
    moves = []
    while len(moves) < max_steps:
        attempt = ao.solve(state)
        if type(attempt) is list:
            moves.extend(attempt)
            return moves
        state, step_moves = attempt
        if not step_moves:
            return None
        moves.extend(step_moves)
    return None
```

## Render videos

Encodes each subfolder's `move_NNN.png` frames into a `video.mp4` via `ffmpeg`, two steps per second.

```python
import subprocess

STEPS_PER_SECOND = 2


def make_video(folder, fps=STEPS_PER_SECOND):
    """Encodes folder's move_NNN.png frames into folder/video.mp4, each frame
    held for 1/fps seconds. Pads to even width/height since libx264 rejects
    odd frame sizes."""
    output_path = os.path.join(folder, "video.mp4")
    subprocess.run(
        [
            "ffmpeg", "-y",
            "-framerate", str(fps),
            "-i", os.path.join(folder, "move_%03d.png"),
            "-vf", "pad=ceil(iw/2)*2:ceil(ih/2)*2",
            "-pix_fmt", "yuv420p",
            output_path,
        ],
        check=True, capture_output=True,
    )
    return output_path


for name in sorted(os.listdir(OUTPUT_DIR)):
    folder = os.path.join(OUTPUT_DIR, name)
    if os.path.isdir(folder):
        make_video(folder)
        print(f"wrote {os.path.join(folder, 'video.mp4')}")
```
